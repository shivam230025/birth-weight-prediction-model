```python
# model training
y= df['bwt']
x= df.iloc[:,1:]
```

```python
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
x_train, x_test, y_train, y_test= tts(x, y, test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge
```

```python
lr= LinearRegression()
ls= Lasso()
rid= Ridge()
```

```python
lr.fit(x_train, y_train)
ls.fit(x_train, y_train)
rid.fit(x_train, y_train)
```

```python
lr_pred= lr.predict(x_test)
ls_pred= ls.predict(x_test)
rid_pred= rid.predict(x_test)
```

```python
#evaluation
from sklearn.metrics import r2_score, mean_squared_error
```

```python
lr_r2score= r2_score(y_test, lr_pred)
ls_r2score= r2_score(y_test, ls_pred)
rid_r2score= r2_score(y_test, rid_pred)
```

```python
lr_mse= mean_squared_error(y_test, lr_pred)
ls_mse= mean_squared_error(y_test, ls_pred)
rid_mse= mean_squared_error(y_test, rid_pred)
```

```python
print("liner regression model evaluation: \n")
print("r2score is : ", lr_r2score)
print("mse is : ", lr_mse)
```

```python
print("lasso model evaluation: \n")
print("r2score is : ", ls_r2score)
print("mse is : ", ls_mse)
```

```python
print("ridge model evaluation: \n")
print("r2score is : ", rid_r2score)
print("mse is : ", rid_mse)
```
